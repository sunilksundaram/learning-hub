## Basics of a Tensor

```python
a = torch.rand(2,3)
b = torch.rand(2,3)
```

```python
print(f'a = {a} \n b = {b}')
```

### Math Operations - add, sub, mul

```python
add = a + b
sub = a - b
mul = a * b
```

```python
print(f'add = {add} \n sub = {sub} \n mul = {mul}')
```

### Tensor Transformations - Transpose

```python
a_transpose = a.T
print(f'a_transpose = {a_transpose}')
```

```python
print(f'a.shape = {a.shape}')
print(f'a_transpose.shape = {a_transpose.shape}')
```

### Tensor Transformations - Matrix Multiplication

```python
dot_AB = a @ b.T
dot_BA = a.T @ b
```

```python
print(f'dot_AB = {dot_AB}\ndot_BA = {dot_BA}')
```

```python
print(f'dot_AB = {dot_AB.shape}\ndot_BA = {dot_BA.shape}')
```

### Broadcasting - Tensor operations of different shapes

```python
a1 = torch.tensor([[1], [2], [3]])
b1 = torch.tensor([4,5,6])
print(f'a1.shape = {a1.shape}\nb1.shape = {b1.shape}')
```

```python
c1 = a1 + b1
print(f'c1.shape = {c1.shape}')
print(f'c1 = {c1}')
```

```python
c2 = b1 + a1
print(f'c2.shape = {c2.shape}')
print(f'c2 = {c2}')
```

### Batch Matrix Multiplication

```python
batch_size = 10
a3 = torch.randn(batch_size, 3, 4) # 10 matrices of size [3,4]
b3 = torch.randn(batch_size, 4, 5) # 10 matrices of size [4,5]
print(f'a3.shape = {a3.shape}')
print(f'a3 = {a3}')
print(f'b3.shape = {b3.shape}')
print(f'b3 = {b3}')
```

```python
c3 = a3 @ b3
```

```python
print(f'c3.shape = {c3.shape}')
print(f'c3 = {c3}')
```

### Batch Matrix Multiplication with Broadcasting

```python
d = torch.randn(4,5) # no batch, so single matrix of [4,5]
e = a3 @ d # with broadcasting
```

```python
print(f'e.shape = {e.shape}')
print(f'e = {e}')
```

#### Data vs Parameter
* __Data__ is simple Tensor with Shape, Type & Device Type.
* __Parameter__ is where autograd keeps track of all operations done on it. Enable using __requires_grad=True__.

```python
x_data = torch.randn(2,2)
print(f'x_data = {x_data} \n of shape: {x_data.shape} \n with data type: {x_data.dtype} \n device: {x_data.device}')
```

```python
w_data = torch.tensor([[1.,2.], [3.,4.]], requires_grad=True)
print(f'w_data = {w_data} \n of shape: {w_data.shape} \n with data type: {w_data.dtype} \n device: {w_data.device}')
```

```python
print(f'x_data: {x_data.requires_grad}')
print(f'w_data: {w_data.requires_grad}')
```

### Tracking the Tensor with Gradient
__Goal:__ Compute z = x * y, where y = a + b

```python
a = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(3.0, requires_grad=True)
x = torch.tensor(4.0, requires_grad=True)
```

```python
y = a + b
z = x * y

print(f'z: {z}')
```

print(f'grad_fn for z: {z.grad_fn}')
print(f'grad_fn for y: {y.grad_fn}')
print(f'grad_fn for a: {a.grad_fn}')

As expected the breadcrumbs (grad_fn) shows that __z__ resulted from a Multiplication, __y__ resulted from an Addition and __a__ was an input (not an operation). When we call __loss.backward()__, PyTorch will walk thru this trail of breadcrumbs to calculate the gradients.

### Reduction - mean, sum, min, max

```python
scores = torch.tensor([[10.,20.,30.],[5.,10.,15.]])
scores_mean = scores.mean() # adds all elements and in this case divides by 6
scores_mean
```

### dim argument - determines which direction to collapse
* dim=0 => collapses the rows, i.e., operates "vertically"
* dim=1 => collapses the columns, i.e., operates "horizontally"

#### Setup - marks of 2 students across 3 subjects

```python
scores1 = torch.tensor([[10.,20.,30.],[5.,10.,15.]])

avg_per_subject = scores1.mean(dim=0) #collapse student dimension
avg_per_student = scores1.mean(dim=1) #collapse subject dimension

print(f'avg_per_subject: {avg_per_subject}')
print(f'avg_per_student: {avg_per_student}')
```

### Dynamic Indexing
Finding index of the highest value

```python
scores_2 = torch.tensor([
    [10, 0, 5, 20, 1],
    [1, 30, 2, 5, 0]
])
```

```python
torch.argmax(scores_2) # gives the index of highest in a single row matrix
```

```python
torch.argmax(scores_2, dim=0) # gives the indexes for each column 
```

```python
torch.argmax(scores_2, dim=1) # gives the index for each row
```

### Torch Gather

```python
data = torch.tensor([
    [10, 11, 12, 13],
    [20, 21, 22, 23],
    [30, 31, 32, 33]
])

indices_to_select = torch.tensor([[2], [0], [3]])
selected_values = torch.gather(data, dim=1, index=indices_to_select)
selected_values
```

```python
indices_to_select1 = torch.tensor([[2], [0], [1]])
torch.gather(data, dim=0, index=indices_to_select1)
```

```python
indices_to_select2 = torch.tensor([[2]])
torch.gather(data, dim=0, index=indices_to_select2)
```
