```python
import torch
```

### Simple Linear Regression
y-hat = X*W+b, where:
* X = input data
* W = weight
* b = bias
* y-hat = model's predicted value

### Setup

```python
# we are using a simple line y = 2x + 1

# Batch size
N = 10

# each data point has 1 feature & 1 output
d_in  = 1
d_out = 1

# create the input data X
X = torch.randn(N, d_in)

true_W = torch.tensor([[2.0]])
true_b = torch.tensor(1.0)
y_true = X @ true_W + true_b + torch.randn(N, d_out) * 0.1 # adding noise

print(f'y_true: {y_true}')
```

### Parameters - the "Brain"

```python
W = torch.randn(d_in, d_out, requires_grad=True)
b = torch.randn(1, requires_grad=True)

print(f'Initial weight: {W}')
print(f'Initial bias  : {b}')
```

### Forward Path

```python
y_hat = X @ W + b

print(f'y_hat: {y_hat}')
```

### Backward Pass

#### Loss - is a number that represents how far is the guess from y_true. Like Mean Squared Error (MSE)

MSE - is the average of all the squared differences of y_hat and y_true

```python
error = y_true - y_hat

print(f'error: {error}')
```

```python
sq_error = error ** 2

print(f'sq_error: {sq_error}')
```

```python
loss = sq_error.mean()

print(f'loss: {loss}')
```

```python
loss.backward()
```

```python
print(f'Gradient for W: {W.grad}')
print(f'Gradient for b: {b.grad}')
```
